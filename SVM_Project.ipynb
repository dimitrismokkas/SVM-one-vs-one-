# **Preprocessing data**

```python
# Importing libraries
import tensorflow as tf
import numpy as np
from sklearn.svm import SVC
from sklearn.decomposition import PCA
from sklearn.model_selection import train_test_split
import time



# Loading dataset
(X_train, y_train), (X_test, y_test) = tf.keras.datasets.cifar10.load_data()

# The classes we will work with
airplane_label = 0
car_label = 1

# Filtering the training dataset so we get only the images from class 0 and 1
train_filtered = (y_train == airplane_label) | (y_train == car_label)
X_train = X_train[train_filtered.reshape(-1)]
y_train = y_train[train_filtered.reshape(-1)]

# Filtering the testing dataset so we get only the images from class 0 and 1
test_filtered = (y_test == airplane_label) | (y_test == car_label)
X_test = X_test[test_filtered.reshape(-1)]
y_test = y_test[test_filtered.reshape(-1)]

# Flattening
X_train_flat = X_train.reshape(len(X_train), -1)
X_test_flat = X_test.reshape(len(X_test), -1)

# Normalizing
X_train_flat = X_train_flat / 255.0
X_test_flat = X_test_flat / 255.0
```

# **SVM Training and Τesting without PCA**

```python
# Training(without PCA)
svm_simple = SVC(kernel='linear')
pca_start_time = time.time()
svm_simple.fit(X_train_flat, y_train.ravel())
pca_end_time = time.time()
```

```python
# Testing(without PCA)
accuracy_simple = svm_simple.score(X_test_flat, y_test)
print(f"Training time without PCA: {pca_end_time - pca_start_time:.2f} seconds")
print(f"Accuracy with PCA: {accuracy_simple:.3f}")
```

# **SVM Training and Τesting with PCA**

```python
# Creating and applying PCA to sets
pca = PCA(n_components=200)
X_train_pca = pca.fit_transform(X_train_flat)
X_test_pca = pca.transform(X_test_flat)
```

```python
# Training(with PCA)
svm_pca = SVC(kernel='rbf')
pca_start_time = time.time()
svm_pca.fit(X_train_pca, y_train.ravel())
pca_end_time = time.time()
```

```python
# Testing(with PCA)
accuracy_pca = svm_pca.score(X_test_pca, y_test)
print(f"Training time with PCA: {pca_end_time - pca_start_time:.2f} seconds")
print(f"Accuracy with PCA: {accuracy_pca:.3f}")
```

# **KNN-1 , KNN - 3 , Centroid (without PCA)**

```python
from tensorflow import keras
from sklearn.neighbors import KNeighborsClassifier, NearestCentroid
from sklearn.metrics import accuracy_score
```

```python
# --- Original Accuracy Calculations(Without PCA) ---
# Creating the Classifiers
knn_1 = KNeighborsClassifier(n_neighbors=1)
knn_3 = KNeighborsClassifier(n_neighbors=3)
centroid = NearestCentroid()

# Training
knn1_start_time = time.time()
knn_1.fit(X_train_flat, y_train.ravel())
knn1_end_time = time.time()
knn1_time = knn1_end_time - knn1_start_time

knn3_start_time = time.time()
knn_3.fit(X_train_flat, y_train.ravel())
knn3_end_time = time.time()
knn3_time = knn3_end_time - knn3_start_time

centroid_start_time = time.time()
centroid.fit(X_train_flat, y_train.ravel())
centroid_end_time = time.time()
centroid_time = centroid_end_time - centroid_start_time


# Testset Prediction
Y_pred_1 = knn_1.predict(X_test_flat)
Y_pred_3 = knn_3.predict(X_test_flat)
Y_pred_centroid = centroid.predict(X_test_flat)

# Accuracy Calculation
accuracy_1 = accuracy_score(y_test, Y_pred_1)
accuracy_3 = accuracy_score(y_test, Y_pred_3)
accuracy_centroid = accuracy_score(y_test, Y_pred_centroid)

# Printing Accuracies
print("Accuracies without PCA:")
print("\n")
print(f"Accuracy of 1-NN: {accuracy_1:.2f}")
print(f"KNN-1 training time: {knn1_time:.2f} seconds")
print("\n")
print(f"Accuracy of 3-NN: {accuracy_3:.2f}")
print(f"KNN-3 training time: {knn3_time:.2f} seconds")
print("\n")
print(f"Accuracy of Nearest Centroid: {accuracy_centroid:.2f}")
print(f"Centroid training time: {centroid_time:.2f} seconds")
print("\n")
print(f"Total training time: {knn1_time + knn3_time + centroid_time:.2f}")

```

# **KNN-1, KNN-3, Centroid (with PCA)**

```python
# --- Original Accuracy Calculations(With PCA) ---


# Creating the Classifiers
knn_1_pca = KNeighborsClassifier(n_neighbors=1)
knn_3_pca = KNeighborsClassifier(n_neighbors=3)
centroid_pca = NearestCentroid()

# Training
knn1_start_time = time.time()
knn_1_pca.fit(X_train_pca, y_train.ravel())
knn1_end_time = time.time()
knn1_time = knn1_end_time - knn1_start_time

knn3_start_time = time.time()
knn_3_pca.fit(X_train_pca, y_train.ravel())
knn3_end_time = time.time()
knn3_time = knn3_end_time - knn3_start_time

centroid_start_time = time.time()
centroid_pca.fit(X_train_pca, y_train.ravel())
centroid_end_time = time.time()
centroid_time = centroid_end_time - centroid_start_time

# Testset Prediction
Y_pred_1_pca = knn_1_pca.predict(X_test_pca)
Y_pred_3_pca = knn_3_pca.predict(X_test_pca)
Y_pred_centroid_pca = centroid_pca.predict(X_test_pca)

# Accuracy Calculation
accuracy_1_pca = accuracy_score(y_test, Y_pred_1_pca)
accuracy_3_pca = accuracy_score(y_test, Y_pred_3_pca)
accuracy_centroid_pca = accuracy_score(y_test, Y_pred_centroid_pca)

# Printing Accuracies
print("Accuracies with PCA:")
print("\n")
print(f"Accuracy of 1-NN: {accuracy_1_pca:.2f}")
print(f"KNN-1 training time: {knn1_time:.2f} seconds")
print("\n")
print(f"Accuracy of 3-NN: {accuracy_3_pca:.2f}")
print(f"KNN-3 training time: {knn3_time:.2f} seconds")
print("\n")
print(f"Accuracy of Nearest Centroid: {accuracy_centroid_pca:.2f}")
print(f"Centroid training time: {centroid_time:.2f} seconds")
print("\n")
print(f"Total training time: {knn1_time + knn3_time + centroid_time:.2f}")
```

# **MLP**

```python
# Create MLP with 1 hidden layer
model = tf.keras.Sequential([
    tf.keras.Input(shape=(X_train_flat.shape[1],)),
    tf.keras.layers.Dense(256, activation='relu'),
    tf.keras.layers.Dense(1)
])
```

```python
# Compile the model
model.compile(optimizer='adam', loss='hinge', metrics=['accuracy'])

# Modify labels to be -1 and +1 (hinge loss expects this)
y_train_hinge = 2 * y_train.flatten() - 1
y_test_hinge = 2 * y_test.flatten() - 1
```

```python
# Training MLP
start_time = time.time()

model.fit(X_train_flat, y_train_hinge, epochs=7, batch_size=64)

end_time = time.time()
training_time = end_time - start_time
print(f"Training time: {training_time:.2f} seconds")
```

```python
# Testing MLP
test_loss, test_accuracy = model.evaluate(X_test_flat, y_test_hinge)
print(f"Test accuracy: {test_accuracy:.3f}")
```
